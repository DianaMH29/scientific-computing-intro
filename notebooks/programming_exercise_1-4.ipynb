Este notebook calcula y compara el eror para el valor "real" y aproximado de $e$.

## 1. Setup
Usaremos la librería `math` para calcular el exponente "real" y `pandas` para organizar los resultados en una tabla. 

```python
import math
import pandas as pd
```

## 2. Definición de variables, ¿Cómo obtendremos cada una? 

- **Valor exacto:** $e$, cáculado con el método definido de math:  `math.exp(1)`.
- **e aproximation:** Usando la fórmula dada: $e=\lim_{n\rightarrow\infty}\left(1+\frac{1}{n}\right)^n$

Y las ya conocidas: 
- **Absolute error:** $E_a = |e - e_aprox|$.
- **Relative error:** $r = E_a / e$

```python
## Definición de funciones: cómputo aproximado de e y cálculo de errores

def get_e_aproximation(n):
    e = (1 + 1/n) ** n
    return e

def get_error_values(n):
    exact = math.exp(1)
    approx = get_e_aproximation(n)
    abs_err = abs(exact - approx)
    rel_err = abs_err / exact
    return exact, approx, abs_err, rel_err
```

## 3. Cálculo y visualización de resultados

- Recorreremos cada $n = 10^k;  k=1,2,...,20$ 
- Se aplicarán las funciones definidas anteriormente
- Se recopilarán los datos y serán formateados como un dataframe de pandas. 

```python
rows = []
for k in range(1, 20):
    n = 10 ** k
    exact, approx, abs_err, rel_err = get_error_values(n)
    rows.append({
        "n": n,
        "k": k,
        "e": exact,
        "e_aprox": round(approx, 10),
        "Abs. Error": round(abs_err, 10),
        "Rel. Error": rel_err,
    })

df = pd.DataFrame(rows)
df
```

## 4. Observaciones: 

- **Error absoluto:** El error tiende a disminuir a medida que crece el valor de $n$, sin embargo, ocurre algo curioso, apartir de $n = 10^10$, el error ansoluto aunmenta a valores incluso mayores que los $n$ más pequeños del rango. Esto es contraintuitivo, pues un n mayor se acerca más a infinito. 
- **Relative error:** Ocurre lo mismo que con el error absoluto, el mejor aciertp (menor error) se obtiene hasta $n = 10^9$, en adelante, el error relativo aumenta. 

¿Qué nos dice esto?, ¿es esta función de aproximación de euler sensible a valores grades de $n$? Tiene cierto sentido, pues si ponemos atención los resultados de la aproximación para n grandes son $\approx 1$, a causa de la sensibildad de la división $1/n$. A valores muy grandes $ \tfrac{1}{n} \approx 0 $, lo que deja la función solo como $1^n$, y por su propiedad, 1 elevado a cualquier número es igual a 1.
